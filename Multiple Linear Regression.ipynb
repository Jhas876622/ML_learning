# Important Libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

# Loading Data 

```python
data = pd.read_csv("multiple_linear_regression.csv")
```

```python
data.head()
```

```python
data.info()
```

# Data Preprocessing

```python
data = data.dropna()
data.shape
```

# Splitting Data 

```python
output_col = "Target"
```

```python
# input
x = data.iloc[:, data.columns  != output_col]
```

```python
x.head()
```

```python
# output
y = data.loc[:, output_col]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.30, random_state = 43)
```

```python
data.shape
```

```python
x_train.shape
```

# <font color = "Dark Red"> Multiple Linear Regression</font>

### Training

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
lr.coef_
```

```python
lr.intercept_
```

### Prediction

```python
pred_value = lr.predict(x_test)
```

### Cost

```python
from sklearn.metrics import mean_squared_error
cost = mean_squared_error(y_test, pred_value)
cost
```

```python
plt.plot(x_test, y_test, "*", color = "green")
plt.plot(x_test,pred_value, "+", color = "red")
plt.title("Performance testing")
plt.xlabel("Input")
plt.ylabel("Output")
```

## ASSUMPTIONS OF LINEAR REGRESSION

#### Homoscedasticity

```python
pred_value
```

```python
residuals = pred_value - y_test
```

```python
residuals
```

```python
plt.scatter(pred_value, residuals)
plt.xlabel("y_pred value")
plt.ylabel("residual value")
plt.show()
```

#### Normality of the Residuals 

```python
sns.displot(residuals)
```
